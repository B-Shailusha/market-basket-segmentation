```python
df["Dt_Customer"]=pd.to_datetime(df["Dt_Customer"],dayfirst = True)
```

```python
df["Age"] = 2025 -df["Year_Birth"]
df["Age"]
```

```python
df["Total_Children"] = df["Kidhome"]+df["Teenhome"]
df["Total_Children"]
```

```python
df.columns
```

```python
spend_cols=['MntWines', 'MntFruits',
       'MntMeatProducts', 'MntFishProducts', 'MntSweetProducts','MntGoldProds']
```

```python
df["Total_Spending"] = df[spend_cols].sum(axis = 1)
df["Total_Spending"]
```

```python
df["Customer_since"] = (pd.Timestamp("today") - df["Dt_Customer"]).dt.days
df["Customer_since"]
```

```python
sns.histplot(df["Age"], bins=30,kde=True)
plt.title("Age Distribution")
plt.show()
```

```python
sns.histplot(df["Income"], bins=30,kde=True)
plt.title("Income Distribution")
plt.show()
```

```python
sns.histplot(df["Total_Spending"], bins=30,kde=True)
plt.title("Total Spending Distribution")
plt.show()
```

```python
sns.boxplot(x="Education",y = "Income", data=df)
plt.xticks(rotation =45)
plt.title("income by education level")
plt.show()
```

```python
sns.boxplot(x="Marital_Status",y = "Total_Spending", data=df)
plt.xticks(rotation =45)
plt.title("spending by marital status")
plt.show()
```

```python
pivot_income=df.pivot_table(values = "Income" , index = "Education" , columns = "Marital_Status")
pivot_income
```

```python
group1=df.groupby("Education")["Total_Spending"].mean().sort_values(ascending=False)
group1
```

```python
group1.plot(kind="bar",color = "skyblue")
plt.ylabel("avg total spending")
plt.xticks(rotation =45)
plt.show()
```

```python
df["AcceptedAny"] = df[["AcceptedCmp1","AcceptedCmp2","AcceptedCmp3","AcceptedCmp4","AcceptedCmp5","Response"]].sum(axis=1)
df["AcceptedAny"].unique()
```

```python
df["AcceptedAny"] = df["AcceptedAny"].apply(lambda x: 1 if x>0 else 0)
df["AcceptedAny"] 
```

```python
group2 = df.groupby("Marital_Status")["AcceptedAny"].mean().sort_values()
group2
```

```python
group2.plot(kind="bar" , color="orange")
plt.ylabel("accepted rate")
plt.xticks(rotation=45)
plt.show()
```

```python
bins = [18,30,40,50,60,70,90]
labels = ["18-29" ,"30-39" ,"40-49" ,"50-59", "60-69" ,"70+"]
df["AgeGroup"] = pd.cut(df["Age"],bins= bins, labels = labels)
df["AgeGroup"] 
```

```python
group3 = df.groupby("AgeGroup")["Income"].mean()
group3
```

```python
group3.plot(kind="bar" , color="green")
plt.xlabel("avg income")
plt.show()
```
